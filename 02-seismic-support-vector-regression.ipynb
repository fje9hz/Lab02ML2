Support Vector Regression (SVR) is similar to SVM, but instead of separating data into classes, it predicts a continuous value. Formulas 1–3 show that SVR allows a small error range, epsilon, and considers predictions within that range acceptable. Formulas 4–5 introduce slack variables and the penalty parameter C, which control how much the model is penalized when points fall outside the accepted range. Formulas 6–9 use Lagrange multipliers to solve the optimization problem and identify support vectors, which are the observations that have the most influence on the model. Finally, Formulas 10–11 use kernel functions to make predictions and allow SVR to capture nonlinear relationships in the data.

```python
import pandas as pd

df = pd.read_excel("data_lab02.xlsx", header=1, skiprows=[2]) #using second row as col. names

features = [
    'B_Lat.', 'B_Long.', 'Height', 'No. of story', 'F1', 'F2',
    'SA1', 'SV1', 'SD1', 'SA2', 'SV2', 'SD2',
    'Avg_Sa', 'Avg_Sv', 'Avg_Sd',
    'E_Lat.', 'E_Long.', 'Magnitude', 'Epicentral distance (R)',
    'PGA', 'PGV', 'PGD', 'AI', 'CAV', 'DP',
    'ZX', 'Effective',
    'Bracketed 1 [0.05g]', 'Bracketed [0.1g]',
    'Bracketed [0.15g]', 'Bracketed [0.2g]',
    'Uniform [0.05g]', 'Uniform [0.1g]',
    'Uniform [0.15g]', 'Uniform [0.2g]',
    'Significant_a1 [(5-75)%]', 'Significant_a2 [(5-95)%]',
    'Significant_v1 [(5-75)%]', 'Significant_v2 [(5-95)%]',
    'Significant_d1 [(5-75)%]', 'Significant_d2 [(5-95)%]'
]

data = df[features + ['Drift']] #new dataset with selected predictors and target variable

data.head()
```

b. Explore the distribution of the target variable (Drift), and create scatterplots between Drift and input features.

```python
import matplotlib.pyplot as plt

# Distribution of Drift
plt.figure()
plt.hist(data['Drift'], bins=30, edgecolor='black')
plt.xlabel('Drift')
plt.ylabel('Frequency')
plt.title('Distribution of Drift')
plt.show()

# Scatterplots between Drift and each predictor
for feature in features:
    plt.figure()
    plt.scatter(data[feature], data['Drift'], alpha=0.5)
    plt.xlabel(feature)
    plt.ylabel('Drift')
    plt.title(f'Drift vs. {feature}')
    plt.show()
```

c. Highlight any particular pattern you observe from the visualizations. Which input features seem potentially more influential in predicting Drift?

After observing the scatterplots, it can be seen that as earthquake-motion intensity increases, Drift generally increases as well. While these relationships are not perfectly linear, the patterns are clearest for variables including SV1, Avg_Sv, PGV, SV2, and PGD. This suggests that these variables may be especially helpful in predicting Drift. Duration and acceleration measures also show some positive relationships, although they appear weaker and more scattered. Overall, the plots suggest that velocity and displacement variables may be more influential in predicting Drift than location variables.
